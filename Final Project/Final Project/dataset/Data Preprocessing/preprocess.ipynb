```python
import json

input_path = "raw_data.jsonl"       # your current file
output_path = "ft_data.jsonl"       # file to use for fine-tuning

with open(input_path, "r", encoding="utf-8") as fin, \
     open(output_path, "w", encoding="utf-8") as fout:
    for line in fin:
        if not line.strip():
            continue
        ex = json.loads(line)
        
        # filter on safety
        if not ex.get("is_safe", False):
            continue

        # build fine-tune record
        record = {
            "input": ex["prompt"],
            "output": ex["response"]
        }
        fout.write(json.dumps(record, ensure_ascii=False) + "\n")
```
